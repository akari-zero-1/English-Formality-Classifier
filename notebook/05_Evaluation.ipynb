###  05 - Đánh giá Tổng hợp tất cả Model

Notebook này so sánh **RoBERTa**, **DistilBERT**, **FastText** trên cùng 1 tập test.

**Tập test dùng:** GYAFC Official Test Set (`gyafc_test_ready.csv`)

| Metric | Ý nghĩa |
|---|---|
| Accuracy | % câu dự đoán đúng |
| Precision | Trong số câu model đoán Formal, bao nhiêu đúng thật |
| Recall | Trong số câu Formal thật, model nhận ra được bao nhiêu |
| F1 Score | Trung bình hài hòa của Precision và Recall |



```python
import pandas as pd
import numpy as np
from sklearn.metrics import (accuracy_score, precision_recall_fscore_support,
                              classification_report, confusion_matrix)
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# ─── Load tap test chinh thuc ───────────────────────────────────────────────
TEST_PATH = 'c:/Users/Admin/Documents/testvscode/formalityClassificationProject/data/gyafc_test_ready.csv'
test_df   = pd.read_csv(TEST_PATH)
y_true    = test_df['label'].tolist()
texts     = test_df['text'].tolist()

print("=== TAP TEST CHINH THUC (GYAFC) ===")
print(f"Tong so cau  : {len(test_df)}")
print(test_df['label'].value_counts().rename({0:'Informal',1:'Formal'}).to_string())
print()
```

##  Đánh giá Model 1: RoBERTa

```python
# ─── Danh gia RoBERTa ────────────────────────────────────────────────────────
from transformers import pipeline
import torch

print("Dang tai RoBERTa...")
roberta_clf = pipeline(
    "text-classification",
    model="c:/Users/Admin/Documents/testvscode/formalityClassificationProject/models/best_roberta_formality",
    tokenizer="c:/Users/Admin/Documents/testvscode/formalityClassificationProject/models/best_roberta_formality",
    device=0 if torch.cuda.is_available() else -1,
    batch_size=32
)

# Du doan theo batch de nhanh hon
print("Dang du doan...")
raw_results = roberta_clf(texts, top_k=1)
y_pred_roberta = [1 if r[0]['label']=='LABEL_1' else 0 for r in raw_results]

# Metrics
acc = accuracy_score(y_true, y_pred_roberta)
p, r, f1, _ = precision_recall_fscore_support(y_true, y_pred_roberta,
                                               average='weighted')
print("\n=== ROBERTA - KET QUA ===")
print(f"Accuracy  : {acc*100:.2f}%")
print(f"Precision : {p*100:.2f}%")
print(f"Recall    : {r*100:.2f}%")
print(f"F1 Score  : {f1*100:.2f}%")
print("\nChi tiet:")
print(classification_report(y_true, y_pred_roberta, target_names=['Informal','Formal']))
```

##  Đánh giá Model 2: DistilBERT

```python
# ─── Danh gia DistilBERT ─────────────────────────────────────────────────────
from transformers import pipeline
import torch

print("Dang tai DistilBERT...")
distilbert_clf = pipeline(
    "text-classification",
    model="c:/Users/Admin/Documents/testvscode/formalityClassificationProject/models/best_distilbert_formality",
    tokenizer="c:/Users/Admin/Documents/testvscode/formalityClassificationProject/models/best_distilbert_formality",
    device=0 if torch.cuda.is_available() else -1,
    batch_size=32
)

print("Dang du doan...")
raw_results = distilbert_clf(texts, top_k=1)
y_pred_distilbert = [1 if r[0]['label']=='LABEL_1' else 0 for r in raw_results]

acc = accuracy_score(y_true, y_pred_distilbert)
p, r, f1, _ = precision_recall_fscore_support(y_true, y_pred_distilbert,
                                               average='weighted')
print("\n=== DISTILBERT - KET QUA ===")
print(f"Accuracy  : {acc*100:.2f}%")
print(f"Precision : {p*100:.2f}%")
print(f"Recall    : {r*100:.2f}%")
print(f"F1 Score  : {f1*100:.2f}%")
print("\nChi tiet:")
print(classification_report(y_true, y_pred_distilbert, target_names=['Informal','Formal']))
```

##  Đánh giá Model 3: FastText

```python
# ─── Danh gia FastText ───────────────────────────────────────────────────────
import os

ft_model_path = 'c:/Users/Admin/Documents/testvscode/formalityClassificationProject/models/best_fasttext_formality.bin'

if not os.path.exists(ft_model_path):
    print("[CANH BAO] Chua tim thay model FastText. Hay chay 04_FastText.ipynb truoc.")
    y_pred_fasttext = None
else:
    try:
        import fasttext
        print("Dang tai FastText model...")
        ft_model = fasttext.load_model(ft_model_path)

        print("Dang du doan...")
        y_pred_fasttext = []
        for text in texts:
            clean = text.replace('\n', ' ')
            pred  = ft_model.predict(clean)
            label = 1 if pred[0][0] == '__label__1' else 0
            y_pred_fasttext.append(label)

        acc = accuracy_score(y_true, y_pred_fasttext)
        p, r, f1, _ = precision_recall_fscore_support(y_true, y_pred_fasttext,
                                                       average='weighted')
        print("\n=== FASTTEXT - KET QUA ===")
        print(f"Accuracy  : {acc*100:.2f}%")
        print(f"Precision : {p*100:.2f}%")
        print(f"Recall    : {r*100:.2f}%")
        print(f"F1 Score  : {f1*100:.2f}%")
        print("\nChi tiet:")
        print(classification_report(y_true, y_pred_fasttext,
                                    target_names=['Informal','Formal']))
    except ImportError:
        print("[CANH BAO] FastText chua duoc cai. Dung Python <= 3.12 de dung model nay.")
        y_pred_fasttext = None
```

##  So sánh Tổng hợp & Biểu đồ

```python
# ─── Bảng so sánh tổng hợp + Confusion Matrix ───────────────────────────────
from sklearn.metrics import accuracy_score, precision_recall_fscore_support, confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

results = {}

for name, preds in [("RoBERTa", y_pred_roberta),
                    ("DistilBERT", y_pred_distilbert),
                    ("FastText", y_pred_fasttext)]:
    if preds is None:
        continue
    acc  = accuracy_score(y_true, preds)
    p, r, f1, _ = precision_recall_fscore_support(y_true, preds, average='weighted')
    results[name] = {"Accuracy":acc*100, "Precision":p*100,
                     "Recall":r*100, "F1":f1*100}

# ── Bảng so sánh ──
summary_df = pd.DataFrame(results).T.round(2)
print("\n" + "="*55)
print("   BANG SO SANH TONG HOP TREN GYAFC OFFICIAL TEST SET")
print("="*55)
print(summary_df.to_string())
print("="*55)

# ── Biểu đồ cột so sánh ──
fig, ax = plt.subplots(figsize=(10, 5))
x = np.arange(len(summary_df.columns))
width = 0.25
colors = ['#4472C4', '#ED7D31', '#A9D18E']

for i, (model_name, row) in enumerate(summary_df.iterrows()):
    ax.bar(x + i*width, row.values, width, label=model_name,
           color=colors[i % len(colors)], alpha=0.85, edgecolor='white')

ax.set_xticks(x + width)
ax.set_xticklabels(summary_df.columns, fontsize=11)
ax.set_ylabel('Score (%)', fontsize=11)
ax.set_title('So sanh Hieu suat 3 Model tren GYAFC Official Test Set',
             fontsize=13, fontweight='bold')
ax.set_ylim(0, 110)
ax.legend(fontsize=10)
ax.grid(axis='y', alpha=0.3)

for bar in ax.patches:
    h = bar.get_height()
    if h > 0:
        ax.text(bar.get_x() + bar.get_width()/2, h + 0.5,
                f'{h:.1f}', ha='center', va='bottom', fontsize=7)
plt.tight_layout()
plt.show()

# ── Confusion Matrix ──
n_models = len(results)
fig, axes = plt.subplots(1, n_models, figsize=(5*n_models, 4))
if n_models == 1:
    axes = [axes]

model_preds = [(n, p) for n, p in [("RoBERTa", y_pred_roberta),
                                    ("DistilBERT", y_pred_distilbert),
                                    ("FastText", y_pred_fasttext)]
               if p is not None]

for ax, (name, preds) in zip(axes, model_preds):
    cm = confusion_matrix(y_true, preds)
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=ax,
                xticklabels=['Informal','Formal'],
                yticklabels=['Informal','Formal'])
    ax.set_title(f'Confusion Matrix\n{name}', fontweight='bold')
    ax.set_xlabel('Du doan')
    ax.set_ylabel('Thuc te')

plt.tight_layout()
plt.show()
```
