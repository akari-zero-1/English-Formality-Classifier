```python
import pandas as pd
import fasttext
import time
import os


DATA_DIR = 'c:/Users/Admin/Documents/testvscode/formalityClassificationProject/data'
MODEL_DIR = 'c:/Users/Admin/Documents/testvscode/formalityClassificationProject/models'
```

```python
print("1. DANG NAP VA CHUYEN DOI DU LIEU...")

# ── Training data: 100% du lieu ──
train_df = pd.read_csv(os.path.join(DATA_DIR, 'gyafc_transformer_ready.csv'))
train_df['fasttext_format'] = '__label__' + train_df['label'].astype(str) + ' ' + train_df['text']
print(f"   Train: {len(train_df)} cau ")

# ── Test data: tap test chinh thuc ──
test_df = pd.read_csv(os.path.join(DATA_DIR, 'gyafc_test_ready.csv'))
test_df['fasttext_format'] = '__label__' + test_df['label'].astype(str) + ' ' + test_df['text']
print(f"   Test : {len(test_df)}  cau ")

# Luu ra file txt (FastText yeu cau format nay)
train_file = os.path.join(DATA_DIR, 'fasttext_train.txt')
test_file  = os.path.join(DATA_DIR, 'fasttext_test.txt')

with open(train_file, 'w', encoding='utf-8') as f:
    f.write('\n'.join(train_df['fasttext_format'].tolist()))

with open(test_file, 'w', encoding='utf-8') as f:
    f.write('\n'.join(test_df['fasttext_format'].tolist()))
```

```python
print("2. BAT DAU HUAN LUYEN FASTTEXT...")
start_time = time.time()

model = fasttext.train_supervised(
    input=train_file,
    epoch=25,
    wordNgrams=2,   # hoc theo cum 2 tu lien tiep
    lr=0.5,
    dim=100
)

elapsed = time.time() - start_time
print(f"   HUAN LUYEN XONG! Thoi gian: {elapsed:.2f} giay")
```

```python
print("\n3. DANH GIA TREN TAP TEST CHINH THUC:")
n, p, r = model.test(test_file)
print(f"   So cau kiem tra : {n}")
print(f"   Precision       : {p * 100:.2f}%")
print(f"   Recall          : {r * 100:.2f}%")

model_path = os.path.join(MODEL_DIR, 'best_fasttext_formality.bin')
model.save_model(model_path)
print(f"\nModel da luu tai: {model_path}")
```
